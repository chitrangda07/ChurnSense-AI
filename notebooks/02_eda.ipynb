```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
path = "../data/raw/Telco_customer_churn.xlsx"
```

```python
DROP_COLUMNS = [
    "CustomerID",
    "Count",
    "Country",
    "State",
    "Lat Long",
    "Churn Reason"
]
def load_data(path):
    df = pd.read_excel(path)
    df = df.drop(columns=DROP_COLUMNS)
    return df
```

```python
df = load_data(path)
```

```python
df.shape
```

```python
df["Churn Label"].value_counts(normalize=True)*100
```

```python
plt.figure(figsize=(6,5))

ax = sns.countplot(
    data=df,
    x="Churn Label",
    hue="Churn Label",
    palette="Set2",
    legend=False
)

for container in ax.containers:
    ax.bar_label(container)

plt.title("Distribution of Customer Churn")
plt.xlabel("Churn Status")
plt.ylabel("Number of Customers")

plt.show()
```

```python
plt.figure(figsize=(6,6))

df["Churn Label"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%",
    explode=[0,0.08],
    shadow=True,
    startangle=90
)

plt.ylabel("")
plt.title("Customer Churn Percentage")

plt.show()
```

```python
def analyze_categorical_feature(df, feature, target="Churn Label"):
    """
    Performs EDA for a categorical feature against the target.
    """

    print("="*60)
    print(f"{feature}")
    print("="*60)

    # Frequency Table
    display(pd.crosstab(df[feature], df[target]))

    # Percentage Table
    display(
        pd.crosstab(
            df[feature],
            df[target],
            normalize="index"
        ) * 100
    )

    # Plot
    plt.figure(figsize=(7,5))

    ax = sns.countplot(
        data=df,
        x=feature,
        hue=target,
        palette="Set2"
    )

    for container in ax.containers:
        ax.bar_label(container)

    plt.title(f"{feature} vs {target}")

    plt.tight_layout()

    plt.show()
```

```python
analyze_categorical_feature(df, "Gender")
```
